```python
# Insert an element at a given position
arr =[2,4,5,6,7,8]
value = 12
position = 3
for i in range(len(arr)-1, position-1, -1):
    arr[i] = arr[i-1]
arr[position] = value
print(arr)
```

```python
# Find the largest and its index
arr = [2,4,5,6,7,8]
largest = arr[0]
for i in range(len(arr)):
    if arr[i] > largest:
        largest = arr[i]
        index = i
print("Largest element is", largest, "at index", index)
```
